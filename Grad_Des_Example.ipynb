```python
# ============================================================
# 1. VERİLERİMİZ
# ============================================================

# Elimizde 4 tane veri noktası var:
#
# (1, 2.1)
# (2, 2.9)
# (3, 3.5)
# (4, 4.6)

x = [1, 2, 3, 4]
y = [2.1, 2.9, 3.5, 4.6]

print("x değerleri:", x)
print("y değerleri:", y)
```

```python
# ============================================================
# 2. MODELİMİZ
# ============================================================

# Kullanacağımız doğru:
#
#        y_pred = w*x + b
#
# w = doğrunun eğimi
# b = doğrunun başlangıç noktası
#
# Başlangıçta w ve b'yi bilmiyoruz.
# Şimdilik ikisini de 0 kabul ediyoruz.

w = 0.0
b = 0.0

print("Başlangıç w:", w)
print("Başlangıç b:", b)
```

```python
# ============================================================
# 3. İLK TAHMİNLERİ HESAPLAYALIM
# ============================================================

# Modelimiz:
#
# y_pred = w*x + b
#
# Şu anda:
#
# w = 0
# b = 0
#
# Bu yüzden her tahmin 0 olacak.

y_pred = []

for xi in x:

    prediction = w * xi + b

    y_pred.append(prediction)

print("Tahminler:", y_pred)
```

```python
# ============================================================
# 4. HATALARI HESAPLAYALIM
# ============================================================

# Hata formülümüz:
#
# error = tahmin - gerçek
#
# yani:
#
# error = y_pred - y

errors = []

for i in range(len(x)):

    error = y_pred[i] - y[i]

    errors.append(error)

print("Hatalar:", errors)
```

```python
# ============================================================
# 5. HATALARIN KARELERİNİ HESAPLAYALIM
# ============================================================

# MSE'de hataların karesini kullanıyoruz:
#
# error^2

squared_errors = []

for error in errors:

    squared_error = error ** 2

    squared_errors.append(squared_error)

print("Hata kareleri:", squared_errors)
```

```python
# ============================================================
# 6. MSE'Yİ HESAPLAYALIM
# ============================================================

# MSE:
#
#             hata karelerinin toplamı
# MSE = -----------------------------------
#                veri sayısı
#
# Yani:
#
# MSE = (1/n) * toplam(error^2)

sum_squared_errors = sum(squared_errors)

number_of_data = len(x)

mse = sum_squared_errors / number_of_data

print("MSE:", mse)
```

```python
# ============================================================
# 7. w İÇİN GRADYAN HESAPLAMA
# ============================================================

# w için gradyanın temel kısmı:
#
#     error * x
#
# Neden?
#
# Çünkü model:
#
#     y_pred = w*x + b
#
# ve w'ye göre türev aldığımızda:
#
#     d(w*x + b) / dw = x
#
# Dolayısıyla her veri noktası için:
#
#     error * x
#
# hesaplıyoruz.

w_gradient_contributions = []

for i in range(len(x)):

    contribution = errors[i] * x[i]

    w_gradient_contributions.append(contribution)

print("w gradyan katkıları:", w_gradient_contributions)
```

```python
# ============================================================
# 8. w GRADYANINI ELDE ETME
# ============================================================

# Bütün veri noktalarının katkılarını topluyoruz.
#
# Çünkü w bütün veri noktaları için ortak bir parametre.

grad_w = sum(w_gradient_contributions)

print("w gradyanı:", grad_w)
```

```python
# ============================================================
# 9. b İÇİN GRADYAN
# ============================================================

# b için:
#
#     grad_b = toplam(error)
#
# Burada error'u x ile çarpmıyoruz.

b_gradient_contributions = []

for error in errors:

    contribution = error

    b_gradient_contributions.append(contribution)

print("b gradyan katkıları:", b_gradient_contributions)
```

```python
# ============================================================
# 10. b GRADYANINI ELDE ETME
# ============================================================

grad_b = sum(b_gradient_contributions)

print("b gradyanı:", grad_b)
```

```python
# ============================================================
# 11. GRADIENT DESCENT İLE w VE b'Yİ GÜNCELLE
# ============================================================

alpha = 0.01

w_new = w - alpha * grad_w

b_new = b - alpha * grad_b

print("Eski w:", w)
print("Yeni w:", w_new)

print("Eski b:", b)
print("Yeni b:", b_new)
```

```python
# ============================================================
# GRADIENT DESCENT
# ============================================================

# Başlangıç değerleri

w = 0.0
b = 0.0

# Öğrenme oranı

alpha = 0.01

# En fazla kaç iterasyon yapacağımız

max_iter = 10000

# Yakınsama için tolerans

precision = 1e-6


for iteration in range(max_iter):

    # --------------------------------------------------------
    # 1. TAHMİNLER
    # --------------------------------------------------------

    y_pred = []

    for xi in x:

        prediction = w * xi + b

        y_pred.append(prediction)


    # --------------------------------------------------------
    # 2. HATALAR
    # --------------------------------------------------------

    errors = []

    for i in range(len(x)):

        error = y_pred[i] - y[i]

        errors.append(error)


    # --------------------------------------------------------
    # 3. w GRADYAN KATKILARI
    # --------------------------------------------------------

    w_gradient_contributions = []

    for i in range(len(x)):

        contribution = errors[i] * x[i]

        w_gradient_contributions.append(contribution)


    # --------------------------------------------------------
    # 4. w GRADYANI
    # --------------------------------------------------------

    grad_w = sum(w_gradient_contributions)


    # --------------------------------------------------------
    # 5. b GRADYAN KATKILARI
    # --------------------------------------------------------

    b_gradient_contributions = []

    for error in errors:

        contribution = error

        b_gradient_contributions.append(contribution)


    # --------------------------------------------------------
    # 6. b GRADYANI
    # --------------------------------------------------------

    grad_b = sum(b_gradient_contributions)


    # --------------------------------------------------------
    # 7. YENİ w VE b
    # --------------------------------------------------------

    w_new = w - alpha * grad_w

    b_new = b - alpha * grad_b


    # --------------------------------------------------------
    # 8. YAKINSAMA KONTROLÜ
    # --------------------------------------------------------

    if (
        abs(w_new - w) < precision
        and
        abs(b_new - b) < precision
    ):

        w = w_new
        b = b_new

        break


    # --------------------------------------------------------
    # 9. YENİ DEĞERLERİ KULLAN
    # --------------------------------------------------------

    w = w_new
    b = b_new


# ============================================================
# SONUÇ
# ============================================================

print("Sonuç")
print("------------")

print("İterasyon:", iteration + 1)

print("w =", w)

print("b =", b)

print(f"Doğru: y = {w:.4f}x + {b:.4f}")
```
